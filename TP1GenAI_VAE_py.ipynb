```python
# =============================================================================
# TUGAS PERSONAL 1 - GENERATIVE AI
# VAE dengan latent space 2 dimensi pada MNIST / Fashion-MNIST
# =============================================================================

# %% CELL 1 -- Setup & konfigurasi
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
from scipy.stats import norm

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ---- KONFIGURASI ----
DATASET = "MNIST"
LATENT_DIM = 2
EPOCHS = 30
BATCH_SIZE = 128
LR = 1e-3
OUTDIR = "hasil_vae"
os.makedirs(OUTDIR, exist_ok=True)

CLASS_NAMES = {
    "MNIST": [str(i) for i in range(10)],
    "FashionMNIST": ["T-shirt", "Trouser", "Pullover", "Dress", "Coat",
                     "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"],
}[DATASET]


# %% CELL 2 -- Load dataset
transform = transforms.ToTensor()   # skala [0,1], cocok untuk BCE loss
DS = datasets.MNIST if DATASET == "MNIST" else datasets.FashionMNIST

train_ds = DS(root="./data", train=True, download=True, transform=transform)
test_ds = DS(root="./data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

print(f"{DATASET}: train={len(train_ds)}, test={len(test_ds)}")


# %% CELL 3 -- Definisi model VAE
class VAE(nn.Module):
    def __init__(self, latent_dim=2, input_dim=784):
        super().__init__()
        self.input_dim = input_dim
        # Encoder: x -> h
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 512), nn.ReLU(),
            nn.Linear(512, 256), nn.ReLU(),
        )
        # Dua kepala: mean dan log-variance dari q(z|x)
        self.fc_mu = nn.Linear(256, latent_dim)
        self.fc_logvar = nn.Linear(256, latent_dim)
        # Decoder: z -> logits x'
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256), nn.ReLU(),
            nn.Linear(256, 512), nn.ReLU(),
            nn.Linear(512, input_dim),          # output logits (belum sigmoid)
        )

    def encode(self, x):
        h = self.encoder(x)
        return self.fc_mu(h), self.fc_logvar(h)

    def reparameterize(self, mu, logvar):
        """Reparameterization trick: z = mu + sigma * eps, eps ~ N(0,I)."""
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        return self.decoder(z)

    def forward(self, x):
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        return self.decode(z), mu, logvar


def vae_loss(logits, x, mu, logvar):
    """ELBO (dinegasikan) = reconstruction loss + KL divergence."""
    recon = F.binary_cross_entropy_with_logits(logits, x, reduction="sum")
    kl = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return recon + kl, recon, kl


model = VAE(LATENT_DIM).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
print(model)
print("Jumlah parameter:", sum(p.numel() for p in model.parameters()))


# %% CELL 4 -- Training loop
history = []

for epoch in range(1, EPOCHS + 1):
    # ---- train ----
    model.train()
    tr_loss = tr_rec = tr_kl = 0.0
    for x, _ in train_loader:
        x = x.view(-1, 784).to(device)
        optimizer.zero_grad()
        logits, mu, logvar = model(x)
        loss, rec, kl = vae_loss(logits, x, mu, logvar)
        loss.backward()
        optimizer.step()
        tr_loss += loss.item(); tr_rec += rec.item(); tr_kl += kl.item()

    n_tr = len(train_ds)
    tr_loss, tr_rec, tr_kl = tr_loss / n_tr, tr_rec / n_tr, tr_kl / n_tr

    # ---- test ----
    model.eval()
    te_loss = te_rec = te_kl = 0.0
    with torch.no_grad():
        for x, _ in test_loader:
            x = x.view(-1, 784).to(device)
            logits, mu, logvar = model(x)
            loss, rec, kl = vae_loss(logits, x, mu, logvar)
            te_loss += loss.item(); te_rec += rec.item(); te_kl += kl.item()
    n_te = len(test_ds)
    te_loss, te_rec, te_kl = te_loss / n_te, te_rec / n_te, te_kl / n_te

    history.append((epoch, tr_loss, te_loss, te_rec, te_kl))
    print(f"Epoch {epoch:02d} | train {tr_loss:7.2f} | test {te_loss:7.2f} "
          f"(recon {te_rec:7.2f} + KL {te_kl:5.2f})")

torch.save(model.state_dict(), f"{OUTDIR}/vae_latent{LATENT_DIM}.pt")

# Kurva loss (opsional tapi bagus buat lampiran)
h = np.array(history)
plt.figure(figsize=(6, 4))
plt.plot(h[:, 0], h[:, 1], label="train loss")
plt.plot(h[:, 0], h[:, 2], label="test loss")
plt.plot(h[:, 0], h[:, 3], "--", label="recon (test)")
plt.plot(h[:, 0], h[:, 4], "--", label="KL (test)")
plt.xlabel("Epoch"); plt.ylabel("Loss per sampel (nats)")
plt.title(f"Kurva training VAE - {DATASET}, latent={LATENT_DIM}")
plt.legend(); plt.grid(alpha=.3); plt.tight_layout()
plt.savefig(f"{OUTDIR}/00_kurva_loss.png", dpi=150)
plt.show()

print("\n>>> CATATAN: final test loss = %.2f (recon %.2f + KL %.2f)"
      % (te_loss, te_rec, te_kl))


# %% CELL 5 -- VISUALISASI 1: Rekonstruksi 10 sampel
model.eval()
x_batch, y_batch = next(iter(test_loader))
x10 = x_batch[:10].view(-1, 784).to(device)
y10 = y_batch[:10]

with torch.no_grad():
    mu, logvar = model.encode(x10)
    recon = torch.sigmoid(model.decode(mu)).cpu()   # mu = rekonstruksi deterministik

fig, axes = plt.subplots(2, 10, figsize=(15, 3.4))
for i in range(10):
    axes[0, i].imshow(x10[i].cpu().view(28, 28), cmap="gray")
    axes[0, i].set_title(CLASS_NAMES[y10[i]], fontsize=9)
    axes[1, i].imshow(recon[i].view(28, 28), cmap="gray")
    axes[0, i].axis("off"); axes[1, i].axis("off")
axes[0, 0].set_ylabel("Input")
fig.text(0.085, 0.72, "INPUT", rotation=90, va="center", fontsize=10)
fig.text(0.085, 0.28, "OUTPUT", rotation=90, va="center", fontsize=10)
plt.suptitle(f"Visualisasi Rekonstruksi - {DATASET} (latent dim = {LATENT_DIM})")
plt.tight_layout()
plt.savefig(f"{OUTDIR}/01_rekonstruksi.png", dpi=150, bbox_inches="tight")
plt.show()


# %% CELL 6 -- VISUALISASI 2: Latent space 2D diwarnai label
zs, ys = [], []
with torch.no_grad():
    for x, y in test_loader:
        mu, _ = model.encode(x.view(-1, 784).to(device))
        zs.append(mu.cpu().numpy()); ys.append(y.numpy())
Z = np.concatenate(zs); Y = np.concatenate(ys)

plt.figure(figsize=(8, 7))
sc = plt.scatter(Z[:, 0], Z[:, 1], c=Y, cmap="tab10", s=3, alpha=.6)
cbar = plt.colorbar(sc, ticks=range(10))
cbar.ax.set_yticklabels(CLASS_NAMES)
plt.xlabel("z[0]"); plt.ylabel("z[1]")
plt.title(f"Latent Space 2D - {DATASET} (10.000 data test)")
plt.grid(alpha=.3); plt.tight_layout()
plt.savefig(f"{OUTDIR}/02_latent_space.png", dpi=150)
plt.show()

# Statistik Bagian B
print("Rentang z[0]: [%.2f, %.2f]  z[1]: [%.2f, %.2f]"
      % (Z[:, 0].min(), Z[:, 0].max(), Z[:, 1].min(), Z[:, 1].max()))
print("\nCentroid tiap kelas (rata-rata posisi di latent space):")
for k in range(10):
    m = Z[Y == k].mean(axis=0)
    s = Z[Y == k].std(axis=0)
    print(f"  {CLASS_NAMES[k]:>10s}: mean=({m[0]:+.2f}, {m[1]:+.2f})  std=({s[0]:.2f}, {s[1]:.2f})")


# %% CELL 7 -- VISUALISASI 3: Grid sampling 10x10
n = 10
# ambil kuantil distribusi normal supaya grid mewakili prior N(0,I)
grid_x = norm.ppf(np.linspace(0.05, 0.95, n))
grid_y = norm.ppf(np.linspace(0.05, 0.95, n))

canvas = np.zeros((n * 28, n * 28))
with torch.no_grad():
    for i, yi in enumerate(grid_y[::-1]):
        for j, xi in enumerate(grid_x):
            z = torch.tensor([[xi, yi]], dtype=torch.float32, device=device)
            img = torch.sigmoid(model.decode(z)).cpu().numpy().reshape(28, 28)
            canvas[i * 28:(i + 1) * 28, j * 28:(j + 1) * 28] = img

plt.figure(figsize=(9, 9))
plt.imshow(canvas, cmap="gray")
plt.xticks(np.arange(n) * 28 + 14, [f"{v:.1f}" for v in grid_x])
plt.yticks(np.arange(n) * 28 + 14, [f"{v:.1f}" for v in grid_y[::-1]])
plt.xlabel("z[0]"); plt.ylabel("z[1]")
plt.title(f"Grid Sampling 10x10 dari Latent Space - {DATASET}")
plt.tight_layout()
plt.savefig(f"{OUTDIR}/03_grid_sampling.png", dpi=150)
plt.show()


# %% CELL 8 -- FAILURE CASES: cari rekonstruksi terburuk
errors, images, labels = [], [], []
with torch.no_grad():
    for x, y in test_loader:
        xf = x.view(-1, 784).to(device)
        mu, _ = model.encode(xf)
        rec = torch.sigmoid(model.decode(mu))
        e = ((rec - xf) ** 2).sum(dim=1)        # squared error per gambar
        errors.append(e.cpu().numpy()); images.append(x.numpy()); labels.append(y.numpy())

errors = np.concatenate(errors)
images = np.concatenate(images)
labels = np.concatenate(labels)

worst = np.argsort(errors)[::-1][:6]
best = np.argsort(errors)[:3]
print("Error rata-rata : %.2f" % errors.mean())
print("Error median    : %.2f" % np.median(errors))
print("\n6 rekonstruksi TERBURUK:")
for r, idx in enumerate(worst, 1):
    print(f"  #{r} index={idx:5d}  label={CLASS_NAMES[labels[idx]]:>10s}  error={errors[idx]:.2f}")
print("\n3 rekonstruksi TERBAIK (buat pembanding):")
for idx in best:
    print(f"  index={idx:5d}  label={CLASS_NAMES[labels[idx]]:>10s}  error={errors[idx]:.2f}")

# error rata-rata per kelas
print("\nError rata-rata per kelas:")
for k in range(10):
    print(f"  {CLASS_NAMES[k]:>10s}: {errors[labels == k].mean():.2f}")

# plot 6 kasus gagal: input vs output + posisi latent-nya
sel = worst
fig, axes = plt.subplots(2, len(sel), figsize=(2.1 * len(sel), 4.6))
with torch.no_grad():
    for i, idx in enumerate(sel):
        xi = torch.tensor(images[idx].reshape(1, 784), device=device)
        mu, _ = model.encode(xi)
        rec = torch.sigmoid(model.decode(mu)).cpu().numpy().reshape(28, 28)
        axes[0, i].imshow(images[idx].reshape(28, 28), cmap="gray")
        axes[0, i].set_title(f"{CLASS_NAMES[labels[idx]]}\nerr={errors[idx]:.0f}", fontsize=9)
        axes[1, i].imshow(rec, cmap="gray")
        axes[1, i].set_title("z=(%.2f, %.2f)" % (mu[0, 0].item(), mu[0, 1].item()), fontsize=8)
        axes[0, i].axis("off"); axes[1, i].axis("off")
plt.suptitle("Failure Case: atas = input asli, bawah = hasil rekonstruksi VAE")
plt.tight_layout()
plt.savefig(f"{OUTDIR}/04_failure_cases.png", dpi=150, bbox_inches="tight")
plt.show()


# %% CELL 9
class PlainAE(nn.Module):
    def __init__(self, latent_dim=2):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 256), nn.ReLU(),
            nn.Linear(256, latent_dim),
        )
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256), nn.ReLU(),
            nn.Linear(256, 512), nn.ReLU(),
            nn.Linear(512, 784),
        )

    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z), z


ae = PlainAE(LATENT_DIM).to(device)
opt_ae = torch.optim.Adam(ae.parameters(), lr=LR)

for epoch in range(1, EPOCHS + 1):
    ae.train()
    tot = 0.0
    for x, _ in train_loader:
        x = x.view(-1, 784).to(device)
        opt_ae.zero_grad()
        logits, _ = ae(x)
        loss = F.binary_cross_entropy_with_logits(logits, x, reduction="sum")
        loss.backward(); opt_ae.step()
        tot += loss.item()
    print(f"[AE] Epoch {epoch:02d} | train recon {tot/len(train_ds):.2f}")

# bandingkan recon loss test AE vs VAE
ae.eval()
ae_rec = 0.0
zs_ae, ys_ae = [], []
with torch.no_grad():
    for x, y in test_loader:
        xf = x.view(-1, 784).to(device)
        logits, z = ae(xf)
        ae_rec += F.binary_cross_entropy_with_logits(logits, xf, reduction="sum").item()
        zs_ae.append(z.cpu().numpy()); ys_ae.append(y.numpy())
ae_rec /= len(test_ds)
Zae = np.concatenate(zs_ae); Yae = np.concatenate(ys_ae)

print("\n>>> PERBANDINGAN")
print(f"  Recon loss test - VAE : {te_rec:.2f}  (+ KL {te_kl:.2f})")
print(f"  Recon loss test - AE  : {ae_rec:.2f}")
print(f"  Rentang latent AE     : z0 [{Zae[:,0].min():.1f}, {Zae[:,0].max():.1f}], "
      f"z1 [{Zae[:,1].min():.1f}, {Zae[:,1].max():.1f}]")
print(f"  Rentang latent VAE    : z0 [{Z[:,0].min():.1f}, {Z[:,0].max():.1f}], "
      f"z1 [{Z[:,1].min():.1f}, {Z[:,1].max():.1f}]")

# gambar perbandingan: latent space & hasil sampling dari N(0,I)
fig, axes = plt.subplots(2, 2, figsize=(12, 11))

axes[0, 0].scatter(Z[:, 0], Z[:, 1], c=Y, cmap="tab10", s=2, alpha=.5)
axes[0, 0].set_title("Latent space VAE (rapat di sekitar 0)")
axes[0, 1].scatter(Zae[:, 0], Zae[:, 1], c=Yae, cmap="tab10", s=2, alpha=.5)
axes[0, 1].set_title("Latent space AE biasa (menyebar bebas)")

# sampling 8x8 dari N(0,I) untuk kedua model
def sample_grid(decode_fn, k=8):
    zz = torch.randn(k * k, LATENT_DIM, device=device)
    with torch.no_grad():
        imgs = torch.sigmoid(decode_fn(zz)).cpu().numpy().reshape(-1, 28, 28)
    cv = np.zeros((k * 28, k * 28))
    for i in range(k):
        for j in range(k):
            cv[i*28:(i+1)*28, j*28:(j+1)*28] = imgs[i*k + j]
    return cv

axes[1, 0].imshow(sample_grid(model.decode), cmap="gray")
axes[1, 0].set_title("Sampling z~N(0,I) lewat decoder VAE")
axes[1, 0].axis("off")
axes[1, 1].imshow(sample_grid(ae.decoder), cmap="gray")
axes[1, 1].set_title("Sampling z~N(0,I) lewat decoder AE")
axes[1, 1].axis("off")

plt.suptitle("VAE vs Autoencoder Biasa (latent dim = 2)")
plt.tight_layout()
plt.savefig(f"{OUTDIR}/05_vae_vs_ae.png", dpi=150)
plt.show()


# %% CELL 10 -- Verifikasi hipotesis latent 10 dimensi
def train_vae(latent_dim, epochs=EPOCHS):
    m = VAE(latent_dim).to(device)
    o = torch.optim.Adam(m.parameters(), lr=LR)
    for _ in range(epochs):
        m.train()
        for x, _ in train_loader:
            x = x.view(-1, 784).to(device)
            o.zero_grad()
            lg, mu, lv = m(x)
            l, _, _ = vae_loss(lg, x, mu, lv)
            l.backward(); o.step()
    m.eval()
    R = K = 0.0
    with torch.no_grad():
        for x, _ in test_loader:
            x = x.view(-1, 784).to(device)
            lg, mu, lv = m(x)
            _, r, k = vae_loss(lg, x, mu, lv)
            R += r.item(); K += k.item()
    return m, R / len(test_ds), K / len(test_ds)

model10, rec10, kl10 = train_vae(10)
print(f"latent=2  -> recon {te_rec:.2f} | KL {te_kl:.2f} | total {te_rec+te_kl:.2f}")
print(f"latent=10 -> recon {rec10:.2f} | KL {kl10:.2f} | total {rec10+kl10:.2f}")

# cek dimensi (KL per dimensi ~0 berarti mati / posterior collapse)
mus, lvs = [], []
with torch.no_grad():
    for x, _ in test_loader:
        mu, lv = model10.encode(x.view(-1, 784).to(device))
        mus.append(mu.cpu().numpy()); lvs.append(lv.cpu().numpy())
MU = np.concatenate(mus); LV = np.concatenate(lvs)
kl_per_dim = (-0.5 * (1 + LV - MU**2 - np.exp(LV))).mean(axis=0)
print("\nKL per dimensi (latent=10):")
for i, v in enumerate(kl_per_dim):
    print(f"  dim {i}: {v:.3f}" + ("   <- hampir mati" if v < 0.05 else ""))

print("\nSemua gambar tersimpan di folder:", OUTDIR)
```
